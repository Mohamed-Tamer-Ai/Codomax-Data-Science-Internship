```python
import pandas as pd
import numpy as np

# --- Objective 2: Import CSV datasets ---
# To make this notebook self-contained, we will first create a sample CSV file programmatically.
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Name': ['Ahmed', 'Sara', 'Ali', 'Mona', 'Omar', 'Nour', 'Hassan', 'Youssef'],
    'Department': ['IT', 'HR', 'IT', 'Marketing', 'IT', 'HR', 'Marketing', 'Finance'],
    'Age': [28, 32, np.nan, 25, 30, 29, np.nan, 40],
    'Salary': [7500, 6000, 8200, 5000, np.nan, 6200, 5400, 9000]
}

# Save as CSV
pd.DataFrame(data).to_csv('employee_data.csv', index=False)
print("Sample 'employee_data.csv' created successfully!\n")

# Now, import the CSV dataset into a Pandas DataFrame
df = pd.read_csv('employee_data.csv')
print("Dataset imported successfully.")
```

```python
# --- Objective 3: Explore the data ---
print("\n--- First 5 rows of the dataset ---")
display(df.head())

print("\n--- Dataset Information ---")
df.info()

print("\n--- Statistical Summary ---")
display(df.describe())
```

```python
# --- Objective 4: Handle missing values and clean data ---
print("\n--- Missing Values Before Cleaning ---")
print(df.isnull().sum())

# Handling Missing Values:
# 1. Fill missing 'Age' with the median age
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)

# 2. Fill missing 'Salary' with the mean salary
mean_salary = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean_salary)

print("\n--- Missing Values After Cleaning ---")
print(df.isnull().sum())

print("\n--- Cleaned Dataset ---")
display(df)
```

```python
# --- Objective 3 & 4: Organize data and apply basic analysis ---

# 1. Grouping Data: Average Salary by Department
print("\n--- Average Salary by Department ---")
avg_salary_dept = df.groupby('Department')['Salary'].mean().round(2).reset_index()
display(avg_salary_dept)

# 2. Filtering & Sorting Data: Employees older than 28, sorted by Salary descending
print("\n--- Employees Older Than 28 (Sorted by Salary) ---")
filtered_df = df[df['Age'] > 28].sort_values(by='Salary', ascending=False)
display(filtered_df)
```
