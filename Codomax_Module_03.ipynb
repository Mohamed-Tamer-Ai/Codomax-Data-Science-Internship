### Objectives Covered:
1. Create charts and graphs using Matplotlib and Seaborn.
2. Build Bar Charts, Line Charts, Pie Charts, Histograms, and Scatter Plots.
3. Identify patterns, trends, and key insights from data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn
sns.set_theme(style="whitegrid")

# Creating a sample dataset for visualization
np.random.seed(42)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [15000, 18000, 16000, 21000, 25000, 23000]
profit = [1500, 1800, 1200, 2500, 3000, 2800]
categories = ['Electronics', 'Clothing', 'Home', 'Beauty', 'Sports']
category_sales = [45000, 30000, 20000, 15000, 8000]

df_time = pd.DataFrame({'Month': months, 'Sales': sales, 'Profit': profit})
df_cat = pd.DataFrame({'Category': categories, 'Sales': category_sales})

# Dataset for Scatter and Histogram
df_customers = pd.DataFrame({
    'Age': np.random.randint(18, 65, 100),
    'Spending_Score': np.random.randint(1, 100, 100)
})

print("Sample datasets created successfully!")
```

### 1. Line Chart (Trend Analysis)

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_time, x='Month', y='Sales', marker='o', color='b', label='Sales')
sns.lineplot(data=df_time, x='Month', y='Profit', marker='s', color='g', label='Profit')
plt.title('Monthly Sales and Profit Trends')
plt.xlabel('Month')
plt.ylabel('Amount (USD)')
plt.legend()
plt.show()

# INSIGHT:
print("Insight: Sales and profit show a general upward trend, peaking in May.")
```

### 2. Bar Chart & Pie Chart (Categorical Analysis)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar Chart
sns.barplot(ax=axes[0], data=df_cat, x='Category', y='Sales', palette='viridis')
axes[0].set_title('Total Sales by Category')
axes[0].set_ylabel('Sales (USD)')

# Pie Chart
axes[1].pie(df_cat['Sales'], labels=df_cat['Category'], autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[1].set_title('Sales Distribution by Category')

plt.tight_layout()
plt.show()

# INSIGHT:
print("Insight: Electronics dominate the sales at 38.1%, while Sports contribute the least.")
```

### 3. Histogram & Scatter Plot (Distribution & Relationships)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
sns.histplot(ax=axes[0], data=df_customers, x='Age', bins=10, kde=True, color='purple')
axes[0].set_title('Customer Age Distribution')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Frequency')

# Scatter Plot
sns.scatterplot(ax=axes[1], data=df_customers, x='Age', y='Spending_Score', color='orange', s=100, alpha=0.7)
axes[1].set_title('Customer Age vs. Spending Score')
axes[1].set_xlabel('Age')
axes[1].set_ylabel('Spending Score')

plt.tight_layout()
plt.show()

# INSIGHT:
print("Insight: Customer ages are fairly evenly distributed. There is no strong linear correlation between Age and Spending Score based on the scatter plot.")
```
