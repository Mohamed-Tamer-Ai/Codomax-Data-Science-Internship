```python
# 1. Variables, Lists, and Dictionaries
print("--- Data Structures ---")

# Variables
intern_name = "Mohamed Tamer Elsheikh"
intern_id = "CDS/INT/2026/26111173"
is_active = True

# List of modules to complete
modules = ["Python Basics", "Data Analysis", "Data Visualization", "Machine Learning"]

# Dictionary storing student data (simulating a simple dataset)
student_scores = {
    "Math": 85,
    "Statistics": 92,
    "Programming": 95,
    "Databases": 88
}

print(f"Intern: {intern_name} | ID: {intern_id}")
print(f"First Module: {modules[0]}")
print(f"Programming Score: {student_scores['Programming']}")
```

```python
# 2. Control Flow (Loops & If/Else) and Functions
print("\n--- Control Flow & Functions ---")

def analyze_scores(scores_dict):
    """
    Function to calculate the average score and determine performance.
    """
    total_score = 0
    num_subjects = len(scores_dict)
    
    # Loop through dictionary
    for subject, score in scores_dict.items():
        print(f"Subject: {subject}, Score: {score}")
        total_score += score
        
    # Calculate average
    average = total_score / num_subjects
    
    # If/Else Condition
    if average >= 90:
        grade = "Excellent (A)"
    elif average >= 80:
        grade = "Very Good (B)"
    else:
        grade = "Good (C)"
        
    return average, grade

# Calling the function
avg_score, final_grade = analyze_scores(student_scores)
print("-" * 20)
print(f"Average Score: {avg_score}")
print(f"Final Grade: {final_grade}")
```
