```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report
from sklearn.datasets import load_iris

# Set plot style
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings('ignore')
```

### 1. Predictability Model 1: Student Score Prediction (Linear Regression)

```python
# 1.1 Create Synthetic Data
np.random.seed(42)
study_hours = np.random.uniform(1, 10, 100)
scores = (study_hours * 9.5) + np.random.normal(0, 5, 100) # True relationship with noise

df_students = pd.DataFrame({'Study_Hours': study_hours, 'Score': scores})

# 1.2 Split the Data
X_stu = df_students[['Study_Hours']]
y_stu = df_students['Score']
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_stu, y_stu, test_size=0.2, random_state=42)

# 1.3 Train the Model
lr_model = LinearRegression()
lr_model.fit(X_train_s, y_train_s)

# 1.4 Evaluate Performance
y_pred_s = lr_model.predict(X_test_s)
print("--- Student Score Prediction Evaluation ---")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_test_s, y_pred_s):.2f}")
print(f"R-squared (R2): {r2_score(y_test_s, y_pred_s):.2f}\n")

# 1.5 Visualize the Regression Line
plt.figure(figsize=(8, 4))
plt.scatter(X_test_s, y_test_s, color='blue', label='Actual Scores')
plt.plot(X_test_s, y_pred_s, color='red', linewidth=2, label='Predicted Regression Line')
plt.title('Study Hours vs. Student Score')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.legend()
plt.show()
```

### 2. Predictability Model 2: House Price Prediction (Multiple Linear Regression)

```python
# 2.1 Create Synthetic Data for House Prices
size = np.random.randint(800, 3500, 100)
bedrooms = np.random.randint(1, 5, 100)
prices = (size * 150) + (bedrooms * 10000) + np.random.normal(0, 20000, 100)

df_house = pd.DataFrame({'Size_sqft': size, 'Bedrooms': bedrooms, 'Price_USD': prices})

# 2.2 Split the Data
X_h = df_house[['Size_sqft', 'Bedrooms']]
y_h = df_house['Price_USD']
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_h, y_h, test_size=0.2, random_state=42)

# 2.3 Train the Model
house_model = LinearRegression()
house_model.fit(X_train_h, y_train_h)

# 2.4 Evaluate Performance
y_pred_h = house_model.predict(X_test_h)
print("--- House Price Prediction Evaluation ---")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_test_h, y_pred_h):.2f}")
print(f"R-squared (R2): {r2_score(y_test_h, y_pred_h):.2f}")
```

### 3. Predictability Model 3: Iris Flower Classification (Random Forest)

```python
# 3.1 Load the real Iris dataset
iris = load_iris()
X_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
y_iris = iris.target

# 3.2 Split the Data
X_train_i, X_test_i, y_train_i, y_test_i = train_test_split(X_iris, y_iris, test_size=0.3, random_state=42)

# 3.3 Train the Classification Model (Random Forest)
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_i, y_train_i)

# 3.4 Evaluate Performance
y_pred_i = rf_model.predict(X_test_i)
print("--- Iris Flower Classification Evaluation ---")
print(f"Accuracy Score: {accuracy_score(y_test_i, y_pred_i) * 100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test_i, y_pred_i, target_names=iris.target_names))
```
