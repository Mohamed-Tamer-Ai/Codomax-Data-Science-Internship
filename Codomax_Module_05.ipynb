### Project Overview: Telco Customer Churn Analysis
**Objectives Covered:**
1. Choose a real-world dataset from Kaggle or public sources (Using the famous Kaggle Telco Customer Churn Dataset).
2. Analyze the dataset and generate meaningful insights.
3. Create visualizations and summarize findings.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings('ignore')

# 1. Load Real-World Dataset (Telco Customer Churn from Kaggle / IBM Public Repo)
# We use a direct raw URL so it runs seamlessly in Colab without needing Kaggle API keys
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)

# Data Preprocessing: Convert TotalCharges to numeric, filling errors with NaN, then dropping NaNs
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(inplace=True)

print("Real-World Dataset Loaded Successfully!")
display(df.head())
```

### Step 1: Exploratory Data Analysis (EDA)

```python
print("\n--- Dataset Info ---")
df.info()

print("\n--- Descriptive Statistics ---")
display(df.describe())

print("\n--- Missing Values Check ---")
print(df.isnull().sum()[df.isnull().sum() > 0]) # Show only columns with missing values if any
```

### Step 2: Data Visualization & Generating Insights

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Churn Distribution (Pie Chart)
churn_counts = df['Churn'].value_counts()
axes[0, 0].pie(churn_counts, labels=churn_counts.index, autopct='%1.1f%%', colors=['#4CAF50', '#F44336'], startangle=90)
axes[0, 0].set_title('Overall Customer Churn Rate')

# 2. Churn by Contract Type (Bar Chart)
sns.countplot(ax=axes[0, 1], data=df, x='Contract', hue='Churn', palette='Set2')
axes[0, 1].set_title('Churn by Contract Type')
axes[0, 1].set_xlabel('Contract Type')
axes[0, 1].set_ylabel('Number of Customers')

# 3. Monthly Charges vs Churn (Box Plot)
sns.boxplot(ax=axes[1, 0], data=df, x='Churn', y='MonthlyCharges', palette='pastel')
axes[1, 0].set_title('Monthly Charges vs Churn')

# 4. Tenure impact on Churn (Histogram/KDE)
sns.histplot(ax=axes[1, 1], data=df, x='tenure', hue='Churn', multiple="stack", kde=True, palette='magma')
axes[1, 1].set_title('Customer Tenure (Months) impact on Churn')

plt.tight_layout()
plt.show()
```

### Step 3: Summary of Findings & Actionable Insights
Based on the Exploratory Data Analysis and Visualizations of the Telco Churn dataset, here are the key insights:

1. **Overall Churn Rate**: The dataset shows an overall churn rate of **26.6%**, indicating that slightly over 1 in 4 customers leave the service.
2. **Contract Type Risk**: Customers on a **Month-to-Month contract** have a significantly higher churn rate compared to those on One-year or Two-year contracts. Long-term contracts effectively lock in customer loyalty.
3. **Price Sensitivity**: The box plot reveals that customers who churned tend to have **higher median Monthly Charges**. This suggests price sensitivity among the customer base.
4. **Tenure & Loyalty**: The histogram shows that the risk of churning is **highest in the first few months** of tenure. If a customer stays beyond 1-2 years, their likelihood of churning drops drastically.

**Actionable Recommendation**: The business should focus on the onboarding experience for new customers and offer targeted incentives (like discounted rates or free upgrades) to convert Month-to-Month users into 1-year contracts, especially those with high monthly bills.
